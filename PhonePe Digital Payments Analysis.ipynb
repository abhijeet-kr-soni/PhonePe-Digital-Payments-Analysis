```python
# Discrepancies between state level and summing up the values from the district level data

state_totals = state_txn.groupby("State").agg(
    State_Transactions=("Transactions", "sum"),
    State_Amount=("Amount (INR)", "sum"),
    State_Registered_Users=("Registered Users", "sum")).reset_index()

district_totals = district_txn.groupby("State").agg(
    District_Transactions=("Transactions", "sum"),
    District_Amount=("Amount (INR)", "sum"),
    District_Registered_Users=("Registered Users", "sum")).reset_index()

comparison = state_totals.merge(district_totals,on="State",how="outer")

numeric_columns = ["State_Transactions", "State_Amount", "State_Registered_Users", 
                   "District_Transactions", "District_Amount", "District_Registered_Users"]

for column in numeric_columns:
    comparison[column] = pd.to_numeric(
        comparison[column], errors="coerce").fillna(0)

comparison["Transaction Difference"] = (
    comparison["State_Transactions"] -
    comparison["District_Transactions"])

comparison["Amount Difference"] = (
    comparison["State_Amount"] -
    comparison["District_Amount"])

comparison["Registered Users Difference"] = (
    comparison["State_Registered_Users"] -
    comparison["District_Registered_Users"])

discrepancies = comparison[
    (comparison["Transaction Difference"] != 0) |
    (comparison["Amount Difference"].abs() > 0.01) |
    (comparison["Registered Users Difference"] != 0)].copy()

display(discrepancies)
```

```python
#  Ratio of each device brand users to the total numbers of registered users in each state

state_device = state_device[["State", "Year", "Quarter", "Brand", "Registered Users"]]

state_txn = state_txn[["State", "Year", "Quarter", "Registered Users"]]

state_device = state_device.rename(columns={"Registered Users": "Device Users"})

state_txn = state_txn.rename(columns={"Registered Users": "Total Registered Users"})

merged = pd.merge(state_device,state_txn,
    on=["State", "Year", "Quarter"],
    how="inner")

merged["Device Usage Ratio"] = (
    merged["Device Users"] /
    merged["Total Registered Users"])

merged["Device Usage Ratio (%)"] = (merged["Device Usage Ratio"] * 100)

result = merged[
    ["State",
     "Year",
    "Quarter",
    "Brand",
    "Device Users",
    "Total Registered Users",
    "Device Usage Ratio",
    "Device Usage Ratio (%)"]]

display(result)
```
